```python
import tensorflow as tf
from tensorflow.keras.preprocessing.image import ImageDataGenerator, load_img, img_to_array

datagen = ImageDataGenerator(
    rotation_range=40,
    width_shift_range=0.4,
    height_shift_range=0.4,
    shear_range=0.4,
    zoom_range=0.4,
    horizontal_flip=True,
    rescale=None,
    fill_mode='wrap'  # nearest,conastant,reflect,wrap
)
```

```python
img= load_img(r"C:\Users\rimsa\Desktop\generative\traning\happy\OIP (20).jpeg")
```

```python
x= img_to_array(img)
x= x.reshape((1,)+ x.shape)

i=0
for batch in datagen.flow(x,batch_size=1,
                          save_to_dir=r"C:\Users\rimsa\Desktop\DATA ARDUMENT"
                          save_prefix="man", save_mode='jpeg'):
i +=1
if i>30:
  break
```
